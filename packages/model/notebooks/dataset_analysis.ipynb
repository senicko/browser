````python
import logging
from joblib import load
from model.tokenizer import StemTokenizer
from model.scraper import load_corpus 
````

````python
corpus = load_corpus()
tokenizer = StemTokenizer()

documents_n = len(corpus)
documents_text = [article.text for article in corpus]
documents_tokens_count = list(map(lambda text: tokenizer.token_count(text), documents_text))

avg_words = sum(documents_tokens_count) / documents_n
max_words = max(documents_tokens_count)
min_words = min(documents_tokens_count)

print("=" * 45)
print("CORPUS TEXT METRICS SUMMARY")
print("=" * 45)
print(f"Total Articles:\t\t{documents_n:>15,}")
print(f"Average Words/Article:\t{avg_words:>15,.2f}")
print(f"Max Words (Longest):\t{max_words:>15,}")
print(f"Min Words (Shortest):\t{min_words:>15,}")
print("=" * 45)
````
